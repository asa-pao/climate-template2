In a bulleted list, how is climate change affecting your home? Include references to citations that support your points, and provide a bibliography at the end of this section:

* Higher energy use. To keep our apartment cool, we have to use more fans (we do not have AC). The increased temperatures increase our housing temperature, and thus energy usage. This is an increasing global challenge (IEA.)
* Increased fire risk. Our little place frequently experiences evacuation warnings due to fire risk. Climate change has been shown to increase the size and aggressiveness of fires (Velev, 2026). With a 1.5°C increase in temperature, we have seen some of the worst fires over the last few years, a significant lack of snow/rainfall, and dryer vegetation (all exponentially increasing the challenge of firefighting).
* Productivity and Stress. Climate change is well-associated with many mental health disorders and tolls (Liu et al., 2021). When our apartment reaches a low of 92 overnight, it is hard to find motivation to cook or workout. Sleep is less consistent, productivity is less, energy is less, etc. Some are directly impacted by heat, other effects are indirectly a result of higher temperatures.
